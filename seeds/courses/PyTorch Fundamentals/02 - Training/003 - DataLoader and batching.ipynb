# Dataset and DataLoader

A `Dataset` answers two questions: how many items (`__len__`) and what is item i (`__getitem__`). A `DataLoader` wraps it with batching, shuffling and parallel workers.

```python
from torch.utils.data import Dataset, DataLoader
class Squares(Dataset):
    def __len__(self):
        return 1000
    def __getitem__(self, i):
        return torch.tensor([float(i)]), torch.tensor([float(i * i)])
loader = DataLoader(Squares(), batch_size=32, shuffle=True, num_workers=2)
```

## Collate and padding

Variable-length sequences need a `collate_fn` that pads to the longest item in the batch and returns an attention mask. Sorting by length before batching (bucketing) cuts wasted padding.

## Pinned memory

`pin_memory=True` lets host-to-GPU copies run asynchronously; combine it with `non_blocking=True` in `.to(device)`.
