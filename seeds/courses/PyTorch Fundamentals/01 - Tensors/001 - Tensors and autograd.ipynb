# Tensors and autograd

A tensor is an n-dimensional array with a dtype and a device. Autograd records every operation on tensors that require gradients and builds a dynamic computation graph.

```python
import torch
x = torch.tensor([2.0, 3.0], requires_grad=True)
y = (x ** 2).sum()
y.backward()
print(x.grad)
```

## The computation graph

`y.backward()` walks the graph from `y` back to every leaf tensor with `requires_grad=True` and accumulates gradients into `.grad`. Gradients accumulate, so call `optimizer.zero_grad()` or `x.grad.zero_()` before the next backward pass.

```python
with torch.no_grad():
    z = x * 2  # no graph is recorded inside no_grad
print(z.requires_grad)
```

## Devices

Move tensors and modules with `.to(device)`. On Apple Silicon the device string is `mps`; on NVIDIA it is `cuda`. All operands of an operation must live on the same device.

```python
device = 'mps' if torch.backends.mps.is_available() else 'cpu'
x = x.to(device)
```
